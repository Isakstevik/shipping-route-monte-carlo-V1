## Part 1 — Container fleet overview
Fuel consumption per day = total fuel consumption / days at sea.

```python
container_df["days_at_sea"] = (
    container_df["Time spent at sea [hours]"] / 24
)
```

```python
container_df["fuel_consumption_per_day"] = (
    container_df["Total fuel consumption [m tonnes]"]
    / container_df["days_at_sea"]
)
```

### Fuel Consumption Data Quality

The MRV 2025 data contains 2,372 container ships. Two screening steps were applied before analysing daily fuel consumption.

**1. Zero time at sea (135 ships removed).**
These ships reported no hours at sea, so fuel consumption per day cannot be calculated. They are typically vessels that only called at EU ports during the period without reporting sea passages within scope. 2,237 ships remain.

**2. Low exposure (112 ships flagged).**
Ships in the lowest 5% of reported time at sea, below about 15.5 days, were flagged. When the number of days is small, the ratio fuel / days becomes sensitive to small reporting differences, such as fuel used in port being spread over very few sea days. These ships are excluded from the analysis sample.

In the 2025 data, the low-exposure group did not contain extreme outliers (highest value about 123 t/day, within the range of the main sample). The filter is therefore a precautionary screening rule, not a correction of obvious errors. The 5% threshold is a statistical choice, not a shipping or regulatory standard.

**Result:** 2,125 ships in the analysis sample. Daily fuel consumption ranges from 4.0 to 170.7 t/day (median 55.1 t/day).

**Effect on the model:** None in practice. The model uses the reported value for the reference vessel, NYK Venus, which had 147 days at sea and is far above the screening threshold. The raw data is kept unchanged in `data/raw/`.

```python
fuel_processed = fuel_analysis_df[
    [
        "IMO Number",
        "Name",
        "Ship type",
        "Total fuel consumption [m tonnes]",
        "Time spent at sea [hours]",
        "days_at_sea",
        "fuel_consumption_per_day"
    ]
].copy()
```

```python
fuel_processed.to_csv(
    "fuel_consumption_processed.csv",
    index = False
)
```

## Part 2 — Reference vessel

Fuel consumption across the container fleet ranges from about 4 to 170 t/day, mainly driven by vessel size. Sampling from the whole fleet would mix feeder vessels with ultra-large ships, so the model instead uses the reported value for one specific reference vessel.

Candidates were three vessels on Asia–North Europe services in the 8,000–9,000 TEU range (NYK Venus, NYK Orion, ONE Hamburg), plus the earlier placeholder vessel CMA CGM EXEMPLARITY for comparison.

```python
candidates = "NYK VENUS|NYK ORION|ONE HAMBURG|EXEMPLARITY"

ships = fuel_analysis_df[fuel_analysis_df["Name"].str.contains(candidates, case=False)]
ships[["IMO Number", "Name", "days_at_sea", "fuel_consumption_per_day"]]
```

```python
nyk_venus = ships.loc[ships["Name"] == "NYK VENUS", "fuel_consumption_per_day"].iloc[0]
share_below = (fuel_analysis_df["fuel_consumption_per_day"] < nyk_venus).mean()
print(f'NYK Venus consumes more than {share_below:.0%} of container ships in the MRV sample')
```

## Findings
- All four candidates consume 73–76 t/day, despite different ages.
- **NYK Venus** (IMO 9312793) is selected: 9,012 TEU, no scrubber (burns VLSFO), deployed on ONE's FE1 Asia–North Europe service.
- Model input: **75.1 t/day at sea**.
- Note: MRV total fuel includes some fuel used in port, so this slightly overstates sea consumption.
- The fleet-wide processed file is kept for reference but is not used by the Monte Carlo model.
